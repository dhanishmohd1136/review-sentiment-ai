```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer

df = pd.read_csv("../../data/processed/processed_reviews.csv")
df.head()
```

```python
X = df['clean_text']
y = df['sentiment']
```

```python
bow_vectoraizer = CountVectorizer()
x_bow = bow_vectoraizer.fit_transform(X)
x_bow.shape

vocabulary = bow_vectoraizer.get_feature_names_out()

print("Number of features :",len(vocabulary))
```

```python
n_gram_vectorizer = CountVectorizer(
    ngram_range=(1,2)
)

x_ngram = n_gram_vectorizer.fit_transform(X)

n_gram_vocabulary = n_gram_vectorizer.get_feature_names_out()

n_gram_vocabulary[:50]
```

```python
bigram_vectorizer = CountVectorizer(
    ngram_range=(2,2)
)

x_bigram = bigram_vectorizer.fit_transform(X)
bi_gram_vocabulary = bigram_vectorizer.get_feature_names_out()

bi_gram_vocabulary[:50]
```

```python
tfidf_vectorizer = TfidfVectorizer(stop_words=None)

X_tfidf = tfidf_vectorizer.fit_transform(X)

tfidf_vocablary = (tfidf_vectorizer.get_feature_names_out())
print(tfidf_vocablary[:50])
```

```python
tfidf_ngram_vectorizer = TfidfVectorizer(
    ngram_range=(1,2),
    stop_words=None

)

x_tfidf_ngram  = tfidf_ngram_vectorizer.fit_transform(X)
```

```python
tfidf_limited = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    stop_words=None
)

X_tfidf_limited = tfidf_limited.fit_transform(X)
```

```python
tfidf_min_df = TfidfVectorizer(
    min_df=2,
    ngram_range=(1, 2),
    stop_words=None
)

X_tfidf_min_df = tfidf_min_df.fit_transform(X)
```

```python
representation_sizes = pd.DataFrame({
    "Representation": [
        "BoW",
        "Unigram + Bigram",
        "Bigram",
        "TF-IDF",
        "TF-IDF + Unigram + Bigram",
        "TF-IDF Limited",
        "TF-IDF min_df=2"
    ],
    "Documents": [
        x_bow.shape[0],
        x_ngram.shape[0],
        x_bigram.shape[0],
        X_tfidf.shape[0],
        x_tfidf_ngram.shape[0],
        X_tfidf_limited.shape[0],
        X_tfidf_min_df.shape[0]
    ],
    "Features": [
        x_bow.shape[1],
        x_ngram.shape[1],
        x_bigram.shape[1],
        X_tfidf.shape[1],
        x_tfidf_ngram.shape[1],
        X_tfidf_limited.shape[1],
        X_tfidf_min_df.shape[1]
    ]
})

representation_sizes
```

```python
comparison = pd.DataFrame({
    "Representation": [
        "BoW Unigram",
        "BoW Unigram + Bigram",
        "TF-IDF Unigram",
        "TF-IDF Unigram + Bigram"
    ],
    "Number of Features": [
        x_bow.shape[1],
        x_ngram.shape[1],
        X_tfidf.shape[1],
        x_tfidf_ngram.shape[1]
    ],
    "Average Non-Zero Features": [
        x_bow.getnnz(axis=1).mean(),
        x_ngram.getnnz(axis=1).mean(),
        X_tfidf.getnnz(axis=1).mean(),
        x_tfidf_ngram.getnnz(axis=1).mean()
    ]
})

comparison
```

```python
def calculate_sparsity(matrix):
    total_elements = matrix.shape[0] * matrix.shape[1]
    non_zero_elements = matrix.nnz

    return (
        1 - non_zero_elements / total_elements
    ) * 100
```

```python
results = pd.DataFrame({
    "Representation": [
        "BoW Unigram",
        "BoW Unigram + Bigram",
        "TF-IDF Unigram",
        "TF-IDF Unigram + Bigram"
    ],
    "Features": [
        x_bow.shape[1],
        x_ngram.shape[1],
        X_tfidf.shape[1],
        x_tfidf_ngram.shape[1]
    ],
    "Average Non-Zero Features": [
        x_bow.getnnz(axis=1).mean(),
        x_ngram.getnnz(axis=1).mean(),
        X_tfidf.getnnz(axis=1).mean(),
        x_tfidf_ngram.getnnz(axis=1).mean()
    ],
    "Sparsity (%)": [
        calculate_sparsity(x_bow),
        calculate_sparsity(x_ngram),
        calculate_sparsity(X_tfidf),
        calculate_sparsity(x_tfidf_ngram)
    ]
})

results
```

```python
sample_index = 0

bow_values = x_bow[sample_index].toarray().ravel()
tfidf_values = X_tfidf[sample_index].toarray().ravel()

value_comparison = pd.DataFrame({
    "term": tfidf_vocablary,
    "bow_value": bow_values,
    "tfidf_value": tfidf_values
})

value_comparison = value_comparison[
    value_comparison["bow_value"] > 0
]

value_comparison.sort_values(
    "tfidf_value",
    ascending=False
).head(20)
```

```python
review_index = 0

bow_review = x_bow[review_index]
tfidf_review = X_tfidf[review_index]

terms = bow_vectoraizer.get_feature_names_out()

review_comparison = pd.DataFrame({
    "term": terms,
    "bow_count": bow_review.toarray().ravel(),
    "tfidf_weight": tfidf_review.toarray().ravel()
})

review_comparison = review_comparison[
    review_comparison["bow_count"] > 0
]

review_comparison.sort_values(
    "tfidf_weight",
    ascending=False
).head(20)
```

```python
mean_tfidf = np.asarray(
    X_tfidf.mean(axis=0)
).ravel()

tfidf_terms = pd.DataFrame({
    "term": tfidf_vocablary,
    "mean_tfidf": mean_tfidf
})

tfidf_terms.sort_values(
    "mean_tfidf",
    ascending=False
).head(20)
```

```python
tfidf_ngram_vocabulary = (
    tfidf_ngram_vectorizer
    .get_feature_names_out()
)

print(
    "Total features:",
    len(tfidf_ngram_vocabulary)
)
```
