```python
df["unique_word_count"] = df["review"].apply(
    lambda text: len(
        set(str(text).lower().split())
    )
)
```

```python
df["word_count"] = df["review"].apply(
    lambda text: len(str(text).split())
)
```

```python
df["character_count"] = df["review"].apply(
    lambda text: len(str(text))
)
```
