```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfVectorizer
)

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv( "../../data/processed/processed_reviews.csv")
```

```python
X_text = df["clean_text"]
y = df["sentiment"]

X_text = df["clean_text"]
y = df["sentiment"]

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


bow_vectorizer = CountVectorizer()

X_train_bow = bow_vectorizer.fit_transform(X_train_text)
X_test_bow = bow_vectorizer.transform(X_test_text)
```

```python
bow_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

bow_model.fit(X_train_bow,y_train)

y_pred_bow = bow_model.predict(X_test_bow)
```

```python
bow_accuracy = accuracy_score(
    y_test,
    y_pred_bow
)

bow_precision = precision_score(
    y_test,
    y_pred_bow,
    pos_label="positive"
)

bow_recall = recall_score(
    y_test,
    y_pred_bow,
    pos_label="positive"
)

bow_f1 = f1_score(
    y_test,
    y_pred_bow,
    pos_label="positive"
)

print("BoW Unigram")
print("=" * 30)

print("Accuracy :", bow_accuracy)
print("Precision:", bow_precision)
print("Recall   :", bow_recall)
print("F1 Score :", bow_f1)
```

```python
print(
    classification_report(
        y_test,
        y_pred_bow
    )
)
```

```python
tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_text)

X_test_tfidf = tfidf_vectorizer.transform(X_test_text)


tfidf_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

tfidf_model.fit(X_train_tfidf,y_train)

y_pred_tfidf = tfidf_model.predict(X_test_tfidf)


tfidf_accuracy = accuracy_score(
    y_test,
    y_pred_tfidf
)

tfidf_precision = precision_score(
    y_test,
    y_pred_tfidf,
    pos_label="positive"
)

tfidf_recall = recall_score(
    y_test,
    y_pred_tfidf,
    pos_label="positive"
)

tfidf_f1 = f1_score(
    y_test,
    y_pred_tfidf,
    pos_label="positive"
)

print("TF-IDF Unigram")
print("=" * 30)

print("Accuracy :", tfidf_accuracy)
print("Precision:", tfidf_precision)
print("Recall   :", tfidf_recall)
print("F1 Score :", tfidf_f1)
```

```python
tfidf_ngram_vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    max_features=20000
)

X_train_tfidf_ngram = (tfidf_ngram_vectorizer.fit_transform(X_train_text))
X_test_tfidf_ngram = (tfidf_ngram_vectorizer.transform(X_test_text))
```

```python
print("Training:",X_train_tfidf_ngram.shape)

print("Testing:",X_test_tfidf_ngram.shape)

tfidf_ngram_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

tfidf_ngram_model.fit(X_train_tfidf_ngram,y_train)

tfidf_ngram_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

tfidf_ngram_model.fit(X_train_tfidf_ngram,y_train)

y_pred_tfidf_ngram = (
    tfidf_ngram_model.predict(
        X_test_tfidf_ngram
    )
)

ngram_accuracy = accuracy_score(
    y_test,
    y_pred_tfidf_ngram
)

ngram_precision = precision_score(
    y_test,
    y_pred_tfidf_ngram,
    pos_label="positive"
)

ngram_recall = recall_score(
    y_test,
    y_pred_tfidf_ngram,
    pos_label="positive"
)

ngram_f1 = f1_score(
    y_test,
    y_pred_tfidf_ngram,
    pos_label="positive"
)

print("TF-IDF Unigram + Bigram")
print("=" * 35)

print("Accuracy :", ngram_accuracy)
print("Precision:", ngram_precision)
print("Recall   :", ngram_recall)
print("F1 Score :", ngram_f1)
```

```python
results = pd.DataFrame({
    "Representation": [
        "BoW Unigram",
        "TF-IDF Unigram",
        "TF-IDF Unigram + Bigram"
    ],
    "Features": [
        X_train_bow.shape[1],
        X_train_tfidf.shape[1],
        X_train_tfidf_ngram.shape[1]
    ],
    "Accuracy": [
        bow_accuracy,
        tfidf_accuracy,
        ngram_accuracy
    ],
    "Precision": [
        bow_precision,
        tfidf_precision,
        ngram_precision
    ],
    "Recall": [
        bow_recall,
        tfidf_recall,
        ngram_recall
    ],
    "F1 Score": [
        bow_f1,
        tfidf_f1,
        ngram_f1
    ]
})

results
```
